## 1. Create a base class called Playlist with a method play() that prints 'Playing playlist'. Then create a subclass SpotifyPlaylist that inherits from Playlist and overrides play() to print 'Playing Spotify playlist'. Call both play() methods to show the difference.

```python
class Playlist:
    def play(self):
        print("Playing playlist")


class SpotifyPlaylist(Playlist):
    def play(self):
        print("Playing Spotify playlist")


p1 = Playlist()
p2 = SpotifyPlaylist()

p1.play()
p2.play()
```

## 2. Build a Product class with a method getDiscount() that returns 5%. Create a subclass called MyntraProduct that overrides getDiscount() to return 15% instead. Instantiate both classes and print the discount for each.<br><br><em><strong>Hint:</strong> Use the same method name in both classes to demonstrate method overriding.</em>

```python
class Product:
    def getDiscount(self):
        return "5%"


class MyntraProduct(Product):
    def getDiscount(self):
        return "15%"


p1 = Product()
p2 = MyntraProduct()

print(f"Product Discount: {p1.getDiscount()}")
print(f"Myntra Discount: {p2.getDiscount()}")
```

## 3. Write a function showProfileDetails(profile) that takes an object of either BasicProfile or InfluencerProfile class and prints the profile's details. Both classes should have a display() method, but InfluencerProfile should override display() to include follower count. Pass both types to showProfileDetails() and observe polymorphic behavior.

```python
class BasicProfile:
    def __init__(self, name, username):
        self.name = name
        self.username = username

    def display(self):
        print(f"Name: {self.name}")
        print(f"Username: {self.username}")


class InfluencerProfile(BasicProfile):
    def __init__(self, name, username, followers):
        super().__init__(name, username)
        self.followers = followers

    def display(self):
        print(f"Name: {self.name}")
        print(f"Username: {self.username}")
        print(f"Followers: {self.followers}")


def showProfileDetails(profile):
    profile.display()


basic = BasicProfile("Aarav", "@aarav")
influencer = InfluencerProfile("Riya", "@riya", 50000)

print("Basic Profile:")
showProfileDetails(basic)

print("\nInfluencer Profile:")
showProfileDetails(influencer)
```

## 4. Imagine you are building a BookMyShow-like ticketing system. Create a base class Ticket with a method getPrice(). Create subclasses MovieTicket and ConcertTicket that override getPrice() to return different prices. Write code that loops through a list of mixed Ticket objects and prints the price for each using polymorphism.<br><br><em><strong>Hint:</strong> Store both MovieTicket and ConcertTicket objects in a single array and call getPrice() in a loop.</em>

```python
class Ticket:
    def getPrice(self):
        return 0


class MovieTicket(Ticket):
    def getPrice(self):
        return 250


class ConcertTicket(Ticket):
    def getPrice(self):
        return 1000


tickets = [MovieTicket(), ConcertTicket()]

for ticket in tickets:
    print(f"Ticket Price: ₹{ticket.getPrice()}")
```
