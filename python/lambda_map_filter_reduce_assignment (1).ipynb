### 1. Add 18% GST to a price

```python
add_gst = lambda price: price * 1.18

price = 500
print(f"Price after GST: ₹{add_gst(price)}")
```

### 2. Clean product names using map()

```python
products = [' pizza ', 'BURGER', ' fries ']
clean_products = list(map(lambda name: name.strip().title(), products))

print(f"Clean product names: {clean_products}")
```

### 3. Find usernames starting with A

```python
usernames = ['Aarav', 'priya', 'Anaya', 'amit', 'Rohan', 'Aisha']
a_names = list(filter(lambda name: name.lower().startswith('a'), usernames))

print(f"Usernames starting with A: {a_names}")
```

### 4. Calculate the total order amount using reduce()

```python
from functools import reduce

order_amounts = [250, 120, 300, 90]
total_bill = reduce(lambda total, amount: total + amount, order_amounts)

print(f"Total bill amount: ₹{total_bill}")
```

### 5. Find the average of ratings 4 and above

```python
from functools import reduce

ratings = [5, 4, 3, 5, 2, 4]
ratings_as_numbers = list(map(lambda rating: rating, ratings))
good_ratings = list(filter(lambda rating: rating >= 4, ratings_as_numbers))
rating_total = reduce(lambda total, rating: total + rating, good_ratings)
average_rating = rating_total / len(good_ratings)

print(f"Ratings of 4 and above: {good_ratings}")
print(f"Average rating: {average_rating:.2f}")
```
