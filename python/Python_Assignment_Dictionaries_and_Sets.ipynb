## 1. Create a Playlist Duration Dictionary

**Question:** Create a dictionary named `playlist_durations` with five playlists and their durations in minutes. Print the duration of one playlist.

````python
playlist_durations = {
    "Daily Mix": 45,
    "Workout Hits": 60,
    "Chill Vibes": 50,
    "Bollywood Favorites": 75,
    "Study Music": 90
}

print(f"Playlist durations: {playlist_durations}")
print(f"Workout Hits duration: {playlist_durations['Workout Hits']} minutes")
````

**Output:**
```text
Playlist durations: {'Daily Mix': 45, 'Workout Hits': 60, 'Chill Vibes': 50, 'Bollywood Favorites': 75, 'Study Music': 90}
Workout Hits duration: 60 minutes
```

## 2. Add a Playlist and Update a Duration

**Question:** Add a new playlist and update the duration of an existing playlist. Print the updated dictionary.

````python
playlist_durations["Morning Motivation"] = 40
playlist_durations["Daily Mix"] = 55

print(f"Updated playlist durations: {playlist_durations}")
````


## 4. Find Common and Unique Food Items

**Question:** Use two sets to find the food items ordered in both weeks (intersection) and all unique items ordered across both weeks (union).

````python
last_week = {"Pizza", "Burger", "Fries", "Pasta"}
this_week = {"Pizza", "Sandwich", "Pasta", "Momos"}

common_items = last_week.intersection(this_week)
all_items = last_week.union(this_week)

print(f"Items ordered last week: {last_week}")
print(f"Items ordered this week: {this_week}")
print(f"Items ordered in both weeks: {common_items}")
print(f"All unique food items: {all_items}")
````

**Output:**
```text
Items ordered last week: {'Pizza', 'Burger', 'Fries', 'Pasta'}
Items ordered this week: {'Pizza', 'Sandwich', 'Pasta', 'Momos'}
Items ordered in both weeks: {'Pizza', 'Pasta'}
All unique food items: {'Pizza', 'Burger', 'Fries', 'Pasta', 'Sandwich', 'Momos'}
```

*Note: Python sets are unordered, so the order of items shown in the output may vary.*

## 5. Find Instagram Users with More Than 10,000 Followers

**Question:** Create a dictionary of Instagram usernames and follower counts. Use a loop to print usernames with more than 10,000 followers.

````python
insta_followers = {
    "travel_diary": 8500,
    "fashion_daily": 15200,
    "foodie_world": 12000,
    "music_lovers": 6400,
    "style_with_me": 24500
}

print("Users with more than 10,000 followers:")

for username, followers in insta_followers.items():
    if followers > 10000:
        print(f"{username}: {followers} followers")
````

**Output:**
```text
Users with more than 10,000 followers:
fashion_daily: 15200 followers
foodie_world: 12000 followers
style_with_me: 24500 followers
```
