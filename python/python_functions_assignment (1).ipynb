### 1. Calculate the final price after discount

```python
def get_final_price(price, discount):
    discount_amount = price * discount / 100
    return price - discount_amount

print(f"Final price: ₹{get_final_price(1000, 20)}")
```

### 2. Format a username

```python
def format_username(name):
    return name.lower().replace(" ", "")

print(f"Username: {format_username('Priya Sharma')}")
```

### 3. Split a bill between people

```python
def split_bill(total_amount, num_people=2):
    return total_amount / num_people

print(f"Each person pays: ₹{split_bill(1000)}")
print(f"Each person pays: ₹{split_bill(1000, 4)}")
```

### 5. Apply a coupon

```python
def apply_coupon(price, coupon=None):
    if coupon == 'SAVE10':
        return price * 0.9
    else:
        return price

print(f"Price with coupon: ₹{apply_coupon(500, 'SAVE10')}")
print(f"Price without coupon: ₹{apply_coupon(500)}")
```
