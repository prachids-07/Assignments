# Python Assignment: Exception Handling

**Topic:** Exception Handling in Python

This notebook contains four simple programs using `try`, `except`, `else`, `finally`, and a custom exception.

## Q1. Safe playlist division

```python
def safe_playlist_divide(total_songs, num_playlists):
    try:
        songs_per_playlist = total_songs // num_playlists
        return songs_per_playlist
    except ZeroDivisionError:
        print("Cannot divide by zero playlists!")

print(safe_playlist_divide(30, 5))
print(safe_playlist_divide(30, 0))
```

## Q2. Bill Splitter

```python
try:
    bill_amount = float(input("Enter bill amount: "))
    num_friends = int(input("Enter number of friends: "))

    if num_friends == 0:
        raise ZeroDivisionError

except ValueError:
    print("Invalid input! Please enter numbers only.")
except ZeroDivisionError:
    print("Number of friends cannot be zero.")
else:
    amount_per_person = bill_amount / num_friends
    print("Amount per person:", round(amount_per_person, 2))
finally:
    print("Thank you for using Bill Splitter!")
```

## Q3. Discount with a custom exception

```python
class InvalidDiscountError(Exception):
    pass

def get_discounted_price(price, discount_percent):
    if discount_percent < 0 or discount_percent > 100:
        raise InvalidDiscountError("Discount must be between 0 and 100%")

    discounted_price = price - (price * discount_percent / 100)
    return discounted_price

try:
    print(get_discounted_price(1000, 20))
    print(get_discounted_price(1000, 120))
except InvalidDiscountError as error:
    print(error)
```

## Q4. Handle invalid Instagram followers input
