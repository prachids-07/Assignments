## 1. Check Spotify Listening Time

**Question:** Ask the user for their Spotify listening time in minutes. If it is above 120 minutes, print `Time for a break!`; otherwise, print `Keep enjoying your music!`.

````python
listening_time = int(input("Enter your Spotify listening time in minutes: "))

if listening_time > 120:
    print(f"Time for a break!")
else:
    print(f"Keep enjoying your music!")
````

```text

Time for a break!
```

**Sample Output 2:**
```text

Keep enjoying your music!
```

## 2. Check Zomato Free Delivery Eligibility

**Question:** Ask the user for their Zomato order amount. If it is above ₹500, print `Eligible for free delivery`; otherwise, print `Delivery charges apply`.

````python
order_amount = float(input("Enter your Zomato order amount in rupees: ₹"))

if order_amount > 500:
    print(f"Eligible for free delivery")
else:
    print(f"Delivery charges apply")
````

Eligible for free delivery
```

**Sample Output 2:**

```

## 3. Apply a Flipkart Discount Rule

**Question:** Ask for the cart value and display the discount message:
- Above ₹2000: 10% discount
- ₹1000 to ₹2000 inclusive: 5% discount
- Below ₹1000: no discount

````python
cart_value = float(input("Enter your Flipkart cart value in rupees: ₹"))

if cart_value > 2000:
    print(f"You get a 10% discount")
elif cart_value >= 1000:
    print(f"You get a 5% discount")
else:
    print(f"No discount")
````

**Sample Output 1:**
```text
Enter your Flipkart cart value in rupees: ₹2500
You get a 10% discount
```

**Sample Output 2:**
```text
Enter your Flipkart cart value in rupees: ₹1500
You get a 5% discount
```

**Sample Output 3:**
```text
Enter your Flipkart cart value in rupees: ₹750
No discount
```

## 4. Classify IPL Fantasy Points

**Question:** Ask the user for their IPL fantasy points and display the correct player category:
- Above 500: `Superstar`
- 300 to 500: `Pro Player`
- 100 to 299: `Casual Player`
- Below 100: `Newbie`

````python
fantasy_points = int(input("Enter your IPL fantasy points: "))

if fantasy_points > 500:
    print(f"Superstar")
elif fantasy_points >= 300:
    print(f"Pro Player")
elif fantasy_points >= 100:
    print(f"Casual Player")
else:
    print(f"Newbie")
````
