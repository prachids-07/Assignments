# String Methods

## 1. Clean a Product Name

**Question:** Take the string `' Redmi-Note 12 Pro '` and remove leading/trailing spaces, convert it to title case, and replace `-` with a space.

````python
product = " Redmi-Note 12 Pro "

cleaned_product = product.strip().replace("-", " ").title()

print(f"Original product name: '{product}'")
print(f"Cleaned product name: {cleaned_product}")
````

**Output:**
```text
Original product name: ' Redmi-Note 12 Pro '
Cleaned product name: Redmi Note 12 Pro
```

## 2. Split Product Features

**Question:** Split the product description into a list of features using `split()` and print each feature on a new line.

````python
description = "OnePlus Nord CE 3 Lite, 8GB RAM, 128GB Storage"

features = description.split(",")

print("Product features:")
for feature in features:
    print(f"- {feature.strip()}")
````

**Output:**
```text
Product features:
- OnePlus Nord CE 3 Lite
- 8GB RAM
- 128GB Storage
```

## 3. Extract the Brand Name

**Question:** Write a function that returns the first word (brand name) from a product name using string indexing and slicing.

````python
def get_brand(product_name):
    space_index = product_name.index(" ")
    return product_name[:space_index]


product_name = "boAt Rockerz 255F Wireless"
brand = get_brand(product_name)

print(f"Product name: {product_name}")
print(f"Brand name: {brand}")
````

**Output:**
```text
Product name: boAt Rockerz 255F Wireless
Brand name: boAt
```

## 5. Replace Spaces and Convert to Uppercase

**Question:** Replace all spaces in the product name with underscores and convert the entire string to uppercase.

````python
product = "realme Narzo 50A Prime"

result = product.replace(" ", "_").upper()

print(f"Original product name: {product}")
print(f"Formatted product name: {result}")
````

**Output:**
```text
Original product name: realme Narzo 50A Prime
Formatted product name: REALME_NARZO_50A_PRIME
```
