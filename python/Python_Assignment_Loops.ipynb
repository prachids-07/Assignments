## 1. Calculate Total Flipkart Order Spending

**Question:** Loop through a list of 10 recent order prices using a `for` loop and print the total amount spent.

````python
order_prices = [499, 1299, 799, 1599, 349, 2199, 899, 599, 1499, 699]
total_amount = 0

for price in order_prices:
    total_amount += price

print(f"Order prices: {order_prices}")
print(f"Total amount spent: ₹{total_amount}")
````

**Output:**
```text
Order prices: [499, 1299, 799, 1599, 349, 2199, 899, 599, 1499, 699]
Total amount spent: ₹10440
```

## 2. Simulate a Zomato Delivery Countdown

**Question:** Use a `while` loop to start a countdown at 10 minutes and print each remaining time until it reaches 0.

````python
remaining_time = 10

while remaining_time >= 0:
    print(f"Time remaining: {remaining_time} minutes")
    remaining_time -= 1

print(f"Your delivery countdown is complete!")
````

**Output:**
```text
Time remaining: 10 minutes
Time remaining: 9 minutes
Time remaining: 8 minutes
Time remaining: 7 minutes
Time remaining: 6 minutes
Time remaining: 5 minutes
Time remaining: 4 minutes
Time remaining: 3 minutes
Time remaining: 2 minutes
Time remaining: 1 minutes
Time remaining: 0 minutes
Your delivery countdown is complete!
```

## 3. Stop Counting Likes with `break`

**Question:** Use a `for` loop and `break` to stop as soon as a post has more than 1,000 likes. Print how many posts were checked.

````python
post_likes = [250, 480, 920, 1350, 760, 1800]
posts_checked = 0

for likes in post_likes:
    posts_checked += 1
    print(f"Checking post {posts_checked}: {likes} likes")

    if likes > 1000:
        print(f"Found a post with more than 1000 likes!")
        break

print(f"Total posts checked: {posts_checked}")
````

**Output:**
```text
Checking post 1: 250 likes
Checking post 2: 480 likes
Checking post 3: 920 likes
Checking post 4: 1350 likes
Found a post with more than 1000 likes!
Total posts checked: 4
```

## 4. Display Spotify Playlists with Their Positions

**Question:** Create a list of eight Spotify playlist names. Use `enumerate()` to print each playlist's position starting from 1.

````python
playlists = [
    "Chill Vibes",
    "Workout Hits",
    "Bollywood Favorites",
    "Study Music",
    "Daily Mix",
    "Morning Motivation",
    "Party Anthems",
    "Relaxing Acoustic"
]

for position, playlist in enumerate(playlists, start=1):
    print(f"{position}. {playlist}")
````

**Output:**
```text
1. Chill Vibes
2. Workout Hits
3. Bollywood Favorites
4. Study Music
5. Daily Mix
6. Morning Motivation
7. Party Anthems
8. Relaxing Acoustic
```
