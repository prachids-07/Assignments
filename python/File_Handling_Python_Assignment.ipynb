## Q1. Create a file and write five song names

```python
with open("my_fav_songs.txt", "w") as f:
    f.write("Kesariya\\n")
    f.write("Heeriye\\n")
    f.write("Apna Bana Le\\n")
    f.write("Perfect\\n")
    f.write("Tum Hi Ho\\n")

print("Songs written successfully.")
```

## Q2. Read the file and print song names in uppercase

```python
with open("my_fav_songs.txt", "r") as f:
    for song in f:
        print(song.strip().upper())
```

## Q3. Read IPL scores from a CSV file and calculate strike rate

Create the sample CSV file below first. The scores are illustrative sample data, not actual IPL match statistics.

```python
import csv

with open("ipl_scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Player", "Runs", "Balls"])
    writer.writerow(["Player A", 45, 30])
    writer.writerow(["Player B", 60, 40])
    writer.writerow(["Player C", 25, 20])
    writer.writerow(["Player D", 50, 25])
    writer.writerow(["Player E", 18, 15])

print("CSV file created successfully.")
```

```python
import csv

with open("ipl_scores.csv", "r") as f:
    reader = csv.reader(f)
    next(reader)

    for row in reader:
        player = row[0]
        runs = int(row[1])
        balls = int(row[2])

        strike_rate = (runs / balls) * 100
        print(player, round(strike_rate, 2))
```

## Q4. Read restaurant details from a JSON file

Create a sample JSON file with the restaurant details, then load it and print names with ratings above 4.0.

```python
import json

restaurants_data = [
    {"name": "Green Bowl", "cuisine": "Indian", "rating": 4.3},
    {"name": "Pasta House", "cuisine": "Italian", "rating": 3.8},
    {"name": "Spice Garden", "cuisine": "Indian", "rating": 4.5},
    {"name": "Burger Point", "cuisine": "Fast Food", "rating": 4.0},
    {"name": "The Food Corner", "cuisine": "Chinese", "rating": 4.2}
]

with open("zomato_restaurants.json", "w") as f:
    json.dump(restaurants_data, f, indent=4)

print("JSON file created successfully.")
```

```python
import json

with open("zomato_restaurants.json", "r") as f:
    restaurants = json.load(f)

for restaurant in restaurants:
    if restaurant["rating"] > 4.0:
        print(restaurant["name"])
```

## Q5. Check whether a file exists using pathlib

```python
from pathlib import Path

file_path = Path("my_fav_songs.txt")

if file_path.exists():
    print(file_path.resolve())
else:
    print("File not found")
```

## Important Points

- `open()` opens a file. `"w"` means write mode and `"r"` means read mode.
- `\\n` moves text to a new line.
- `strip()` removes surrounding whitespace, and `upper()` converts text to uppercase.
- `csv.reader()` reads CSV rows.
- Strike rate = (Runs / Balls) × 100.
- `json.load()` reads JSON data from a file.
- `Path.exists()` checks whether a file exists, and `Path.resolve()` returns its absolute path.
