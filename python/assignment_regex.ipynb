```python
# Q1. Use re.findall() to extract all Indian phone numbers in the format +91-XXXXXXXXXX from a given text.
import re
text = """
Contact: +91-9876543210
Office: +91-8765432109
Invalid number: 9876543210
"""
pattern = r'\+91-\d{10}'
phone_numbers = re.findall(pattern, text)
print("Indian Phone Numbers:", phone_numbers)
```

```python
# Q2. Use re.search() to find the first date in DD/MM/YYYY format from a multiline string.
import re
text = """
The first event is on 15/08/2026.
The second event is on 26/01/2027.
"""
pattern = r'\d{2}/\d{2}/\d{4}'
result = re.search(pattern, text)
if result:
    print("First Date:", result.group())
else:
    print("No date found.")
```

```python
# Q3. Use re.sub() to replace all numbers (both integer and decimal) in a product review text with the word 'NUM' and print the updated text.
import re
review = "This product costs 499.99 rupees and has a rating of 4.5 out of 5. It comes with 2 years warranty."
pattern = r'\d+(?:\.\d+)?'
updated_text = re.sub(pattern, 'NUM', review)
print("Updated Review:")
print(updated_text)
```

```python
# Q4. Create a function extract_usernames(text) that finds all email addresses in a given text and returns only the usernames (the part before @).
import re
def extract_usernames(text):
    pattern = r'([A-Za-z0-9._%+-]+)@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'
    usernames = re.findall(pattern, text)
    return usernames
text = "Contact us at prachi@gmail.com or support123@yahoo.com and admin@company.in"
result = extract_usernames(text)
print("Usernames:", result)
```

```python
# Q5. Write a regular expression pattern that matches both Indian (+91-XXXXXXXXXX) and US (XXX) XXX-XXXX phone numbers.
import re
text = """
Indian number: +91-9876543210
US number: (123) 456-7890
Another Indian number: +91-8765432109
"""
pattern = r'\+91-\d{10}|\(\d{3}\) \d{3}-\d{4}'
result = re.findall(pattern, text)
print("Phone Numbers:", result)
```
