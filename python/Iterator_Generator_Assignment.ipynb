## Q1. Use iter() and next() to manually loop through a list of your top 5 favorite songs from Spotify, printing each song name one by one.

```python
songs = ["Love You Zindagi", "Ilahi", "Mardani", "Safarnama", "Zinda"]

song_iterator = iter(songs)

print("My Top 5 Spotify Songs:")
print(next(song_iterator))
print(next(song_iterator))
print(next(song_iterator))
print(next(song_iterator))
print(next(song_iterator))
```

## Q2. Create a program that uses enumerate() to print each item in a Zomato food order list along with its position number (starting from 1).

```python
food_order = ["Pizza", "Burger", "French Fries", "Sandwich", "Cold Coffee"]

for position, item in enumerate(food_order, start=1):
    print(f"{position}. {item}")
```

## Q3. Given two lists — one of cricket teams and one of their scores in the latest IPL match — use zip() to print each team with its score in the format: 'Team: Score'.

```python
teams = ["CSK", "MI", "RCB", "KKR", "GT"]
scores = [185, 172, 198, 190, 165]

for team, score in zip(teams, scores):
    print(f"{team}: {score}")
```

## Q4. Write a generator function called playlist_generator that takes a list of YouTube video titles and yields the next video title each time next() is called.

```python
def playlist_generator(video_titles):
    for title in video_titles:
        yield title


youtube_videos = [
    "Python Tutorial",
    "Machine Learning Basics",
    "Web Development",
    "Data Science",
    "Python Projects"
]

playlist = playlist_generator(youtube_videos)

print(next(playlist))
print(next(playlist))
print(next(playlist))
print(next(playlist))
print(next(playlist))
```

## Q5. Build a generator function called discount_coupon_generator that yields a new coupon code each time it is called, starting from 'SAVE10', 'SAVE20', 'SAVE30', and so on up to 'SAVE50'.

```python
def discount_coupon_generator():
    for number in range(10, 51, 10):
        yield f"SAVE{number}"


coupons = discount_coupon_generator()

print(next(coupons))
print(next(coupons))
print(next(coupons))
print(next(coupons))
print(next(coupons))
```
