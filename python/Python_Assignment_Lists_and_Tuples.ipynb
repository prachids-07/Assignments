# Lists and Tuples

## 1. Update a Playlist

**Question:** Create a list called `playlist` containing three songs, use `append()` to add one more song, and print the updated list.

````python
playlist = ["Love You Zindagi", "Ilahi", "Kar Har Maidaan Fateh"]

print(f"Original playlist: {playlist}")

playlist.append("Zinda")

print(f"Updated playlist: {playlist}")
````

**Output:**
```text
Original playlist: ['Love You Zindagi', 'Ilahi', 'Kar Har Maidaan Fateh']
Updated playlist: ['Love You Zindagi', 'Ilahi', 'Kar Har Maidaan Fateh', 'Zinda']
```

## 2. Add Items to a Shopping Cart

**Question:** Start with `cart = ['T-shirt', 'Jeans']`, use `extend()` to add shoes and a watch, and print the final cart.

````python
cart = ["T-shirt", "Jeans"]

print(f"Cart before adding items: {cart}")

cart.extend(["Shoes", "Watch"])

print(f"Final shopping cart: {cart}")
````

**Output:**
```text
Cart before adding items: ['T-shirt', 'Jeans']
Final shopping cart: ['T-shirt', 'Jeans', 'Shoes', 'Watch']
```

## 3. Try to Update a Tuple

**Question:** Create a tuple of three favorite clothing brands and try to change the second brand. Observe the error and explain why it occurs in a comment.

````python
fav_brands = ("Nike", "Adidas", "Puma")

print(f"Favorite brands: {fav_brands}")

try:
    fav_brands[1] = "Zara"
except TypeError as error:
    print(f"Error: {error}")

# A tuple is immutable, which means its items cannot be changed after creation.
# Use a list instead when you need to add, remove, or update items.
````

**Output:**
```text
Favorite brands: ('Nike', 'Adidas', 'Puma')
Error: 'tuple' object does not support item assignment
```

**Explanation:** Tuples are immutable, so an item cannot be reassigned after the tuple is created.

## 4. Remove the Last Order

**Question:** Create a list of Zomato orders, use `pop()` to remove the last item, and print the removed item and the updated list.

````python
orders = ["Pizza", "Burger", "Fries", "Pasta"]

print(f"Order history before removal: {orders}")

removed_order = orders.pop()

print(f"Removed order: {removed_order}")
print(f"Updated order history: {orders}")
````

**Output:**
```text
Order history before removal: ['Pizza', 'Burger', 'Fries', 'Pasta']
Removed order: Pasta
Updated order history: ['Pizza', 'Burger', 'Fries']
```

## 5. IPL Teams and Season Years

**Question:** Create a list of IPL team names and a tuple of season years. Add a team to the list and try to add a year to the tuple. Explain when to use a list versus a tuple in a comment.

````python
ipl_teams = ["Chennai Super Kings", "Mumbai Indians", "Rajasthan Royals"]
season_years = (2022, 2023, 2024)

print(f"IPL teams before update: {ipl_teams}")
ipl_teams.append("Royal Challengers Bengaluru")
print(f"IPL teams after adding a team: {ipl_teams}")

print(f"IPL season years: {season_years}")

try:
    season_years = season_years + (2025,)
    print(f"Updated season years: {season_years}")
except TypeError as error:
    print(f"Error: {error}")

# Lists are useful for data that may change, such as adding an IPL team.
# Tuples are useful for a fixed collection of values.
# A tuple cannot be changed in place, but a new tuple can be created by combining tuples.
````

**Output:**
```text
IPL teams before update: ['Chennai Super Kings', 'Mumbai Indians', 'Rajasthan Royals']
IPL teams after adding a team: ['Chennai Super Kings', 'Mumbai Indians', 'Rajasthan Royals', 'Royal Challengers Bengaluru']
IPL season years: (2022, 2023, 2024)
Updated season years: (2022, 2023, 2024, 2025)
```

**Explanation:** A list is mutable, so its items can be added or removed. A tuple is immutable, so you cannot modify it in place. You can, however, create a new tuple by combining it with another tuple.
