# Python Data Types, Type Conversion, Functions and Strings

This notebook demonstrates:
1. Declaring variables with different Python data types and using `type()`.
2. Converting strings to floats and calculating a total.
3. Creating and testing a function with integer and float inputs.
4. Converting an integer phone number to a string and masking all but the last four digits.

## 1. Variables and Data Types

Declare four variables for age, GPA, name, and Spotify subscription status. Print each variable along with its data type using the `type()` function.

```python
# Declaring variables with different data types
age = 18
gpa = 8.5
name = "Prachi"
spotifySubscription = True

# Printing each variable and its type
print("Age:", age, "Type:", type(age))
print("GPA:", gpa, "Type:", type(gpa))
print("Name:", name, "Type:", type(name))
print("Spotify Subscription:", spotifySubscription, "Type:", type(spotifySubscription))
```

## 2. Converting Price Strings to Floats

The given prices are stored as strings. Each price is converted to a float using `float()` inside a loop, and then all prices are added to calculate the total.

```python
# List of prices stored as strings
prices = ['199.99', '349.50', '120.00']

total = 0.0

# Convert each price from string to float and add it to total
for price in prices:
    total += float(price)

print("Total sum:", total)
```

## 3. Offer Eligibility Function

The function `is_offer_eligible(amount)` returns `True` when the amount is greater than or equal to 500. Otherwise, it returns `False`. The function is tested with both integer and float values.

```python
# Function to check whether an amount is eligible for an offer
def is_offer_eligible(amount):
    if amount >= 500:
        return True
    else:
        return False

# Testing with an integer input
print("Integer input (500):", is_offer_eligible(500))

# Testing with a float input
print("Float input (499.99):", is_offer_eligible(499.99))
```

## 4. Masking a Phone Number

Convert the phone number from an integer to a string and use string slicing to keep only the last four digits. The remaining digits are replaced with asterisks.

```python
# Phone number entered as an integer
phoneNumber = 9876543210

# Convert the phone number to a string
phoneString = str(phoneNumber)

# Keep only the last four digits and mask the rest
maskedPhone = "*" * (len(phoneString) - 4) + phoneString[-4:]

print("Masked phone number:", maskedPhone)
```
