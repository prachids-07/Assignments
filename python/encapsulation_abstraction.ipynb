## 1. Create a class named Song on your favorite music app that has a private attribute _play_count. Add methods to set and get the play count value.

```python
class Song:
    def __init__(self):
        self.__play_count = 0

    def set_play_count(self, count):
        self.__play_count = count

    def get_play_count(self):
        return self.__play_count


song = Song()

song.set_play_count(150)
print(f"Play Count: {song.get_play_count()}")
```

## 2. Write a MovieTicket class for a BookMyShow-style app with a private _price attribute. Implement get_price() and set_price() methods to safely access and update the ticket price.

```python
class MovieTicket:
    def __init__(self):
        self.__price = 0

    def set_price(self, price):
        self.__price = price

    def get_price(self):
        return self.__price


ticket = MovieTicket()

ticket.set_price(250)
print(f"Ticket Price: ₹{ticket.get_price()}")
```

## 3. Build a MyntraProduct class with a private _discount attribute. Add a method apply_discount() that sets the discount, and a method get_final_price() that returns the price after applying the discount.<br><br><em><strong>Hint:</strong> Use getters and setters to control access to the _discount and _price attributes.</em>

```python
class MyntraProduct:
    def __init__(self, price):
        self.__price = price
        self.__discount = 0

    def set_discount(self, discount):
        self.__discount = discount

    def apply_discount(self, discount):
        self.set_discount(discount)

    def get_final_price(self):
        final_price = self.__price - (self.__price * self.__discount / 100)
        return final_price


product = MyntraProduct(2000)

product.apply_discount(20)

print("Original Price: ₹2000")
print("Discount: 20%")
print(f"Final Price: ₹{product.get_final_price()}")
```

## 4. Create a class called InstaStory with a private _viewer_count attribute. Write a method to increment the viewer count each time a new viewer watches the story, and a getter to retrieve the current count.

```python
class InstaStory:
    def __init__(self):
        self.__viewer_count = 0

    def increment_viewer(self):
        self.__viewer_count += 1

    def get_viewer_count(self):
        return self.__viewer_count


story = InstaStory()

story.increment_viewer()
story.increment_viewer()
story.increment_viewer()

print(f"Total Viewers: {story.get_viewer_count()}")
```
