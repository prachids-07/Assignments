# Recursion & Scope

## 1. Sum of the first n array elements

```python
def sumArray(arr, n):
    if n == 0:
        return 0
    return arr[n - 1] + sumArray(arr, n - 1)

play_counts = [120, 340, 550, 210, 400]
print(f"Total plays for the first 3 songs: {sumArray(play_counts, 3)}")
```

## 2. Reverse a string using recursion

The function processes one character at a time without using `reverse()` or slicing.

```python
def reverseString(s):
    def reverse_from(index):
        if index < 0:
            return ""
        return s[index] + reverse_from(index - 1)
    return reverse_from(len(s) - 1)

username = "priya"
print(f"Reversed username: {reverseString(username)}")
```

## 3. Local and global scope

```python
discount = 15

def applyCoupon():
    discount = 10
    print(f"Discount inside the function: {discount}%")

applyCoupon()
print(f"Discount outside the function: {discount}%")
```

## 4. Total likes for the first n posts

```python
def countLikes(posts, n):
    if n == 0:
        return 0
    return posts[n - 1] + countLikes(posts, n - 1)

post_likes = [120, 340, 550, 280, 410]
print(f"Total likes for the first 3 posts: {countLikes(post_likes, 3)}")
```

## 5. Sum of digits and the call stack

```python
def sum_digits(number):
    number = abs(number)
    print(f"Calling sum_digits({number})")

    if number < 10:
        print(f"Base case: returning {number}")
        return number

    last_digit = number % 10
    remaining = number // 10
    result = last_digit + sum_digits(remaining)
    print(f"Returning {last_digit} + sum_digits({remaining}) = {result}")
    return result

print(f"Sum of digits: {sum_digits(1234)}")
```

### How the call stack works

Each recursive call waits on the call stack until the next call returns. For `1234`, the function calls itself with `123`, then `12`, then `1`. The base case returns `1`. The waiting calls then add the remaining digits: `2 + 1 = 3`, `3 + 3 = 6`, and `4 + 6 = 10`. The final answer is `10`.
