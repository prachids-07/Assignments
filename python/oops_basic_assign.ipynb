## 1. Define a Python class called Song with attributes title, artist, and duration (in seconds), and create an object representing your favorite song.

```python
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration


favorite_song = Song("Love You Zindagi", "Amit Trivedi", 230)

print("Song:", favorite_song.title)
print("Artist:", favorite_song.artist)
print("Duration:", favorite_song.duration, "seconds")
```

## 2. Add a method play_preview() to your Song class that prints 'Playing 30 second preview of [title] by [artist]'. Create two Song objects and call play_preview() on both

```python
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration

    def play_preview(self):
        print(f"Playing 30 second preview of {self.title} by {self.artist}")


song1 = Song("Love You Zindagi", "Amit Trivedi", 230)
song2 = Song("Ilahi", "Arijit Singh", 230)

song1.play_preview()
song2.play_preview()
```

## 3. Create a class called FoodOrder with attributes restaurant_name, items (a list), and order_total. Write an __init__ constructor that sets these attributes, then create an object for a Zomato-style order with at least two items.

```python
class FoodOrder:
    def __init__(self, restaurant_name, items, order_total):
        self.restaurant_name = restaurant_name
        self.items = items
        self.order_total = order_total


my_order = FoodOrder("The Pizza Hub", ["Margherita Pizza", "Garlic Bread"], 350)

print("Restaurant:", my_order.restaurant_name)
print("Items:", my_order.items)
print("Order Total: Rs.", my_order.order_total)
```

## 4. Add a method add_item(self, item, price) to your FoodOrder class that adds the item to the items list and updates order_total. Demonstrate by adding two items to your order object and printing the updated total.<br><br><em><strong>Hint:</strong> Use self.items.append(item) and update self.order_total inside the method.</em>

```python
class FoodOrder:
    def __init__(self, restaurant_name, items, order_total):
        self.restaurant_name = restaurant_name
        self.items = items
        self.order_total = order_total

    def add_item(self, item, price):
        self.items.append(item)
        self.order_total += price


my_order = FoodOrder("The Pizza Hub", ["Margherita Pizza", "Garlic Bread"], 350)

print("Before adding items:")
print("Items:", my_order.items)
print("Order Total: Rs.", my_order.order_total)

my_order.add_item("French Fries", 120)
my_order.add_item("Cold Coffee", 80)

print("\nAfter adding items:")
print("Items:", my_order.items)
print("Updated Order Total: Rs.", my_order.order_total)
```
