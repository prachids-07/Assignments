## Q4. Flipkart Product Ratings
Write a Python function that takes a matrix representing Flipkart product ratings (each row: [product_id, user_id, rating]) and returns a list of all ratings for a given product_id.

```python
def get_ratings(matrix, product_id):
    ratings = []

    for row in matrix:
        if row[0] == product_id:
            ratings.append(row[2])

    return ratings

ratings_matrix = [
    [101, 1, 4.5],
    [102, 2, 4.0],
    [101, 3, 5.0],
    [103, 4, 3.5],
    [101, 5, 4.0]
]

print("Ratings:", get_ratings(ratings_matrix, 101))
```
