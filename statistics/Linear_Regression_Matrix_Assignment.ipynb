## Q1. Matrix Multiplication

Matrix A is a 2×3 matrix and matrix B is a 3×2 matrix. Their product is a 2×2 matrix.

```python
# Q1. Matrix multiplication

A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8],
    [9, 10],
    [11, 12]
]

product = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(3):
            product[i][j] = product[i][j] + A[i][k] * B[k][j]

print("Matrix A:", A)
print("Matrix B:", B)
print("Product matrix:", product)
```

`Product matrix: [[58, 64], [139, 154]]`

## Q2. Slope and Intercept of the Best-Fit Line

The step counts are used as x-values and calories burned as y-values.

Formula:
- Mean = Sum of values / Number of values
- Slope: m = Σ((x - x̄)(y - ȳ)) / Σ((x - x̄)²)
- Intercept: c = ȳ - m x̄

The best-fit line is y = mx + c.

```python
# Q2. Calculate slope and intercept without libraries

steps = [3000, 4500, 5000, 6500, 7000, 8500, 10000]
calories = [120, 160, 180, 220, 240, 280, 320]

n = len(steps)

mean_steps = sum(steps) / n
mean_calories = sum(calories) / n

covariance_sum = 0
variance_sum = 0

for i in range(n):
    covariance_sum = covariance_sum + (steps[i] - mean_steps) * (calories[i] - mean_calories)
    variance_sum = variance_sum + (steps[i] - mean_steps) ** 2

m = covariance_sum / variance_sum
c = mean_calories - (m * mean_steps)

print("Mean steps:", mean_steps)
print("Mean calories:", mean_calories)
print("Slope (m):", m)
print("Intercept (c):", c)
print("Best-fit line: y =", m, "x +", c)
```

## Q3. Predict Calories

This function uses the calculated slope and intercept from Q2.

```python
# Q3. Predict calories

def predict_calories(steps):
    predicted = m * steps + c
    return predicted

steps_input = int(input("Enter number of steps: "))
prediction = predict_calories(steps_input)

print("Predicted calories burned:", prediction)
```

## Q4. Real-World Example of a Best-Fit Line

A best-fit line could be useful in an app like Spotify. For example, Spotify can collect information about how long a user listens to music and how often the user likes or skips songs.
